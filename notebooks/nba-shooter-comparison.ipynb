## Part 1 - NBA Player Comparision

```python
# As usual, we begin by importing the packages we will need

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# The data consists of the shot log for the NBA season 2016/17

shot = pd.read_csv(r"data/NBA Shotlog_16_17.csv")
pd.set_option('display.max_columns', 100)
print(shot.columns.tolist())
shot.describe().round(2)
shot.current_shot_outcome.unique()
```

```python
# A simple plot of coordinates

x = shot['location_x']
y = shot['location_y']
plt.scatter(x,y, s=.005,c='r', marker= '.')
```

```python
Durant = shot[shot['shoot_player']=='Kevin Durant']
Howard = shot[shot['shoot_player']=='Dwight Howard']
Jordan = shot[shot['shoot_player']=='DeAndre Jordan']
Westbrook = shot[shot['shoot_player']=='Russell Westbrook']
BLopez = shot[shot['shoot_player']=='Brook Lopez']
RLopez = shot[shot['shoot_player']=='Robin Lopez']
```

## All Shooter Comparision

```python
fig, (axL, axR, ax3, ax4, ax5, ax6) = plt.subplots(1, 6, figsize=(94/6, 50/6), sharex=True, sharey=True)

# --- Durant (LEFT) ---
cL = np.where(Durant['current_shot_outcome'].eq('SCORED'), 'r',
     np.where(Durant['current_shot_outcome'].eq('MISSED'), 'b', 'g'))
axL.scatter(Durant['location_x'], Durant['location_y'], s=10, c=cL, marker='.')
axL.set_xlim(500, 950)
axL.set_ylim(0, 500)
axL.grid(True)
axL.set_title("Kevin Durant", fontsize=15)

# --- Westbrook (RIGHT) ---
cR = np.where(Westbrook['current_shot_outcome'].eq('SCORED'), 'r',
     np.where(Westbrook['current_shot_outcome'].eq('MISSED'), 'b', 'g'))
axR.scatter(Westbrook['location_x'], Westbrook['location_y'], s=10, c=cR, marker='.')
axR.set_xlim(500, 950)
axR.set_ylim(0, 500)
axR.grid(True)
axR.set_title("Russell Westbrook", fontsize=15)

# --- Westbrook (RIGHT) ---
c3 = np.where(Jordan['current_shot_outcome'].eq('SCORED'), 'r',
     np.where(Jordan['current_shot_outcome'].eq('MISSED'), 'b', 'g'))
ax3.scatter(Jordan['location_x'], Jordan['location_y'], s=10, c=c3, marker='.')
ax3.set_xlim(500, 950)
ax3.set_ylim(0, 500)
ax3.grid(True)
ax3.set_title("DeAndre Jordan", fontsize=15)

# --- Westbrook (RIGHT) ---
c4 = np.where(Howard['current_shot_outcome'].eq('SCORED'), 'r',
     np.where(Howard['current_shot_outcome'].eq('MISSED'), 'b', 'g'))
ax4.scatter(Howard['location_x'], Howard['location_y'], s=10, c=c4, marker='.')
ax4.set_xlim(500, 950)
ax4.set_ylim(0, 500)
ax4.grid(True)
ax4.set_title("Dwight Howard", fontsize=15)

# --- Brook Lopez (RIGHT) ---
c5 = np.where(BLopez['current_shot_outcome'].eq('SCORED'), 'r',
     np.where(BLopez['current_shot_outcome'].eq('MISSED'), 'b', 'g'))
ax5.scatter(BLopez['location_x'], BLopez['location_y'], s=10, c=c5, marker='.')
ax5.set_xlim(500, 950)
ax5.set_ylim(0, 500)
ax5.grid(True)
ax5.set_title("Broook Lopez", fontsize=15)

# --- Robin Lopez (RIGHT) ---
c6 = np.where(RLopez['current_shot_outcome'].eq('SCORED'), 'r',
     np.where(RLopez['current_shot_outcome'].eq('MISSED'), 'b', 'g'))
ax6.scatter(RLopez['location_x'], RLopez['location_y'], s=10, c=c6, marker='.')
ax6.set_xlim(500, 950)
ax6.set_ylim(0, 500)
ax6.grid(True)
ax6.set_title("Robin Lopez", fontsize=15)

plt.tight_layout()
plt.show()
```
